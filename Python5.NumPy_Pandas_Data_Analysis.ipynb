## NumPy Array Operations

**Problem statement:** To learn how to use NumPy for creating arrays, performing basic array operations, and conducting numerical computations through analysing temperature data.

**Scenario:** You are given a list of daily average temperatures (in degrees Celsius) recorded over two weeks. Your task is to use NumPy to analyse this data.

### Step 1: Create a 1D NumPy Array

**1.1** Create a 1D NumPy array named `temperatures_w1` for Week 1 with the following values: `[22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9]`

```python
temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])

print(temperatures_w1)
```

`np.array()` turns a Python list into a NumPy array. The array holds the seven daily temperatures for Week 1, one value per day.

### Step 2: Inspection and Properties

**2.1** Inspect the shape, data type, and the number of elements in the `temperatures_w1` array.

```python
print("Shape            :", temperatures_w1.shape)
print("Data type        :", temperatures_w1.dtype)
print("Number of elements:", temperatures_w1.size)
```

The shape `(7,)` means a one-dimensional array with 7 items. The data type is `float64` because the temperatures have decimals, and `size` confirms there are 7 elements.

### Step 3: Array Operations

**3.1** Convert `temperatures_w1` from Celsius to Fahrenheit using the formula: `Fahrenheit = (Celsius * 9/5) + 32`

```python
temperatures_f = (temperatures_w1 * 9/5) + 32

print("Celsius   :", temperatures_w1)
print("Fahrenheit:", temperatures_f)
```

NumPy applies the formula to every element at once, so no loop is needed. The original Celsius array is not changed, and the converted values are stored in a new array called `temperatures_f`.

**3.2** Find the maximum, minimum, and mean temperatures for the week.

```python
print("Maximum:", temperatures_w1.max())
print("Minimum:", temperatures_w1.min())
print("Mean   :", round(temperatures_w1.mean(), 2))
```

`max()`, `min()` and `mean()` work directly on the array. The mean is rounded to 2 decimal places to keep the output tidy. These values are in degrees Celsius.

### Step 4: Array Slicing and Indexing

**4.1** Extract the temperatures for the first three days of the week.

```python
first_three_days = temperatures_w1[:3]

print(first_three_days)
```

`[:3]` takes the items from the start up to, but not including, position 3. That gives positions 0, 1 and 2, which are days 1 to 3.

**4.2** Extract the temperatures for the weekend (last two days).

```python
weekend = temperatures_w1[-2:]

print(weekend)
```

A negative index counts from the end of the array, so `[-2:]` takes the last two values.

**4.3** Extract the temperatures for the middle three days of the week.

```python
middle_three_days = temperatures_w1[2:5]

print(middle_three_days)
```

With 7 days, the middle three are days 3, 4 and 5. In Python counting from 0, that is positions 2, 3 and 4, so the slice is `[2:5]`.

### Step 5: Create a 2D Array

**5.1** Create a 2D NumPy array named `temperatures` where each row represents temperatures for a different week:
- Row 1: `[22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9]` (Week 1)
- Row 2: `[19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]` (Week 2)

```python
temperatures = np.array([
    [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],   # Week 1
    [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]    # Week 2
])

print(temperatures)
```

A list of two lists becomes a 2D array. Each row is one week and each column is one day of the week.

### Step 6: Inspect and Slice the 2D Array

**6.1** Inspect and print the shape, data type, and total number of elements in the `temperatures` array.

```python
print("Shape             :", temperatures.shape)
print("Data type         :", temperatures.dtype)
print("Number of elements:", temperatures.size)
```

The shape `(2, 7)` means 2 rows (weeks) and 7 columns (days). The array holds 2 x 7 = 14 elements in total, all stored as `float64`.

**6.2** Extract the temperatures for each week and the weekends (last two days) for both weeks.

```python
week1 = temperatures[0]
week2 = temperatures[1]
weekends = temperatures[:, -2:]

print("Week 1   :", week1)
print("Week 2   :", week2)
print("Weekends (both weeks):")
print(weekends)
```

`temperatures[0]` and `temperatures[1]` pick out one full row each. In `temperatures[:, -2:]`, the `:` before the comma means every row, and `-2:` after it means the last two columns. The result has one row per week, showing Saturday and Sunday.


## 🎓 Part 2: Pandas Series

### Step 1: Creating a Pandas Series

**1.1** Create a Pandas Series named `marks` with the following data:
- Marks: 95, 92, 89, 85, 80
- Custom indices: `'Rank1'`, `'Rank2'`, `'Rank3'`, `'Rank4'`, `'Rank5'`

```python
marks = pd.Series([95, 92, 89, 85, 80],
                  index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5'])

print(marks)
```

A Series is a one-dimensional labelled array. Here the labels are `Rank1` to `Rank5` instead of the default 0 to 4, so each mark can be found by the rank it belongs to.

### Step 2: Indexing and Slicing

**2.1** Use the integer index position to access the mark of the 1st rank student.

```python
first_rank_mark = marks.iloc[0]

print("Mark of the 1st rank student:", first_rank_mark)
```

Position 0 is the first item in the Series, so `marks.iloc[0]` gives the 1st rank student's mark. Older tutorials write `marks[0]` for this, but in current Pandas versions a number inside square brackets is treated as a label when the index is text, so `iloc` is the safe way to use a position.

**2.2** Use the `loc` accessor to retrieve the marks of the top 3 ranks by specifying their index labels.

```python
top_three = marks.loc['Rank1':'Rank3']

print(top_three)
```

`loc` works with labels. Unlike a normal Python slice, a `loc` slice includes the last label, so `'Rank1':'Rank3'` returns Rank1, Rank2 and Rank3.

**2.3** Use the `iloc` accessor to retrieve the mark of the 3rd rank student.

```python
third_rank_mark = marks.iloc[2]

print("Mark of the 3rd rank student:", third_rank_mark)
```

`iloc` works with positions that start from 0, so the 3rd student is at position 2.

**2.4** Apply a boolean mask to filter the Series and retrieve the ranks where the marks are greater than 90.

```python
mask = marks > 90

print("Boolean mask:")
print(mask)
print()
print("Marks above 90:")
print(marks[mask])
print()
print("Ranks with marks above 90:", list(marks[mask].index))
```

`marks > 90` gives a True or False value for each rank. Passing that mask back into the Series keeps only the True rows. The ranks themselves are the index labels of the filtered result.

### Step 3: Manipulating Series

**3.1** Modify the mark of the student with the 1st rank to 100.

```python
marks.loc['Rank1'] = 100

print(marks)
```

Assigning to a label with `loc` replaces that value in place. Rank1 now has 100, and the other marks are unchanged.

**3.2** Remove the entry corresponding to the last rank from the Series.

```python
marks = marks.drop('Rank5')

print(marks)
```

`drop()` returns a new Series without the given label, so the result is saved back into `marks`. Rank5 is now gone and four entries remain.

**3.3** Compute the CGPA by dividing each mark in the Series by 10.

```python
cgpa = marks / 10

print(cgpa)
```

Dividing a Series by a number divides every value, and the labels stay the same. The result is stored in a new Series called `cgpa`, so the marks are kept as they are.


## 🛒 Part 3: Pandas DataFrame

### Step 1: Creating a Pandas DataFrame

**1.1** Create a Pandas DataFrame named `transactions` with the following data:
- TransactionID: 101, 102, 103, 104, 105, 106, 107, 108, 109, 110
- ProductCategory: `'Electronics'`, `'Clothing'`, `'Electronics'`, `'Furniture'`, `'Clothing'`, `'Electronics'`, `'Furniture'`, `'Clothing'`, `'Furniture'`, `'Electronics'`
- Region: `'North'`, `'South'`, `'North'`, `'East'`, `'West'`, `'North'`, `'East'`, `'West'`, `'South'`, `'North'`
- Amount: 200, 150, 300, 450, 200, 250, 300, 180, 350, 400

```python
transactions = pd.DataFrame({
    'TransactionID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'ProductCategory': ['Electronics', 'Clothing', 'Electronics', 'Furniture', 'Clothing',
                        'Electronics', 'Furniture', 'Clothing', 'Furniture', 'Electronics'],
    'Region': ['North', 'South', 'North', 'East', 'West',
               'North', 'East', 'West', 'South', 'North'],
    'Amount': [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
})

print(transactions)
```

A dictionary where each key is a column name and each value is a list of that column's data becomes a DataFrame. Each row is one transaction.

### Step 2: Data Exploration

**2.1** Display the `transactions` DataFrame and its basic information, including head, tail, shape, column names, and data types.

```python
print("DataFrame:")
print(transactions)

print("\nHead (first 5 rows):")
print(transactions.head())

print("\nTail (last 5 rows):")
print(transactions.tail())

print("\nShape:", transactions.shape)
print("\nColumn names:", list(transactions.columns))

print("\nData types:")
print(transactions.dtypes)
```

`head()` and `tail()` show the first and last five rows by default. The shape `(10, 4)` means 10 rows and 4 columns. The three text columns are stored as strings, and the ID and amount columns are integers.

**2.2** Display only the `'ProductCategory'` and `'Amount'` columns.

```python
print(transactions[['ProductCategory', 'Amount']])
```

Passing a list of column names inside square brackets returns just those columns. The double brackets are needed because the inner brackets create the list.

**2.3** Retrieve the last 3 columns of the DataFrame using `iloc` or `loc`.

```python
print(transactions.iloc[:, -3:])
```

In `iloc[:, -3:]`, the `:` before the comma selects all rows and `-3:` selects the last three columns, which are `ProductCategory`, `Region` and `Amount`.

**2.4** Filter rows where the `'Region'` is `'North'` and `'Amount'` is greater than 200.

```python
north_high = transactions[(transactions['Region'] == 'North') & (transactions['Amount'] > 200)]

print(north_high)
```

Each condition goes inside its own brackets and the two are joined with `&`, which means both must be true. Of the four North transactions, three have an amount above 200.

**2.5** Find the value counts for the `'ProductCategory'` column.

```python
print(transactions['ProductCategory'].value_counts())
```

`value_counts()` counts how many times each category appears, with the most frequent first. Electronics has the most transactions.

**2.6** Find the unique values in the `'Region'` column.

```python
print(transactions['Region'].unique())
```

`unique()` lists each distinct value once, in the order it first appears. There are four regions.

**2.7** Group by `'Region'` and find the mean amount for each region.

```python
print(transactions.groupby('Region')['Amount'].mean())
```

`groupby('Region')` splits the rows by region, and `['Amount'].mean()` averages the amount within each group. The regions are shown in alphabetical order.

### Step 3: Manipulating the DataFrame

**3.1** Modify the `'Amount'` for the transaction with `TransactionID` 102 to 165.

```python
transactions.loc[transactions['TransactionID'] == 102, 'Amount'] = 165

print(transactions)
```

The first part inside `loc` finds the row where the ID is 102, and the second part picks the `Amount` column. Only that one cell changes, from 150 to 165.

**3.2** Add a new column `'Discount'` by calculating 10% of the `'Amount'`.

```python
transactions['Discount'] = transactions['Amount'] * 0.10

print(transactions)
```

Assigning to a column name that does not exist yet creates it. Multiplying by 0.10 works on the whole column at once, so every row gets its own discount.

**3.3** Remove the row with `TransactionID` 109.

```python
transactions = transactions.drop(transactions[transactions['TransactionID'] == 109].index)

print(transactions)
```

The row with ID 109 is found first, and `drop()` removes it by its index label. The remaining rows keep their original index numbers, so the number 8 is missing from the index. That is normal, and the data itself is correct.

**3.4** Delete the `'Discount'` column from the DataFrame.

```python
transactions = transactions.drop(columns='Discount')

print(transactions)
```

`drop(columns=...)` removes the column and the DataFrame goes back to its four original columns. The edit to transaction 102 and the removal of transaction 109 are still in place.
